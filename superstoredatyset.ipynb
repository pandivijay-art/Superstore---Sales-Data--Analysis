### 1.which catagory has the higesst dales

```python
highest_sales=superstore_data.groupby( by="Category",)["Sales"].sum().sort_values(ascending=False)
highest_sales
```

```python
highest_sales.plot(kind ="bar")
py.title("Higest Sales Category")
py.xlabel("Category")
py.ylabel("Sales")
py.show()
```

## insight from  first question 
###  The Technology category recorded the highest total sales in the dataset, indicating that it generated the highest sales revenue among all categories.

### 2. which region generate the highest sales

```python
superstore_data["Region"]
```

```python
region_sales=superstore_data.groupby("Region")["Sales"].sum().sort_values(ascending=False)
```

```python
region_sales
```

### 2.1 Data visualization chart

```python
import matplotlib.pyplot as py
```

```python
py.bar(x=region_sales.index,height=region_sales.values,)
py.title("Higest Sales Region")
py.xlabel("Region")
py.ylabel("Sales")
py.show()
```

### insight from  question 2:


#### The West region recorded the highest total sales in the dataset. The company can ensure sufficient inventory and continue its successful sales strategies in the West region. The South region recorded the lowest sales, so the company can explore marketing campaigns and sales improvement strategies to increase sales in that region.

### 3.How do sales change over time

```python
superstore_data["month"]=superstore_data["Order Date"].dt.month
```

```python
superstore_data["month"]
```

```python
monthly_sales=superstore_data.groupby("month")["Sales"].sum().reset_index()
```

```python
type(monthly_sales)
```

### Data visualization 

```python
import seaborn as se
```

```python
import matplotlib.pyplot as py
```

```python
se.lineplot(x="month",y="Sales",data=superstore_data,markers="O")
py.title("How sales change over time.")
py.xlabel("Month")
py.ylabel("Sales")
py.xticks(range(1,13))
py.show()
```

#### Insight from question 3

#### Sales increased significantly during the 3rd month and remained relatively high around the following period. After that, sales showed a declining trend during the middle months before recovering later in the year. This indicates that sales performance was not consistent throughout the year

### 4.which sub category has the highest Sales

```python
superstore_data["Sub-Category"]
```

```python
sub_category=superstore_data.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False).reset_index()
```

```python
sub_category
```

### Data visualization 

```python
se.barplot(data=sub_category, x="Sub-Category",y="Sales",hue="Sub-Category",)
py.xticks(rotation=90)
py.tight_layout()
```

### insight from question 4

#### Phones recorded the highest total sales among all sub-categories. On the other hand, Labels, Fasteners, and Art recorded the lowest sales. This shows that sales performance varies significantly across different sub-categories.

### 5.which segment  has the highest Sales?
